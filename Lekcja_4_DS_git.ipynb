```python
# Zadanke 13
pomiary_cisnienia= [120, 125, 118, 190, 122]

srednie_cisnienie = sum(pomiary_cisnienia) / len(pomiary_cisnienia)
print(srednie_cisnienie)

cisnienie_kwadraty = [(pomiar - srednie_cisnienie)**2 for pomiar in pomiary_cisnienia]
odchylenie_standardowe_cisnienie = (sum(cisnienie_kwadraty)/len(cisnienie_kwadraty))**0.5
print(odchylenie_standardowe_cisnienie)

dolna_granica = srednie_cisnienie - 3 * odchylenie_standardowe_cisnienie
gorna_granica = srednie_cisnienie + 3 * odchylenie_standardowe_cisnienie

for pomiar in pomiary_cisnienia:
    if dolna_granica > pomiar > gorna_granica:
        status = "ANOMALIA"
    else: 
        status = 'W NORMIE'
    print(f"Ciśnienie {pomiar} - {status}")
```

```python
# Zadanie 15
wagi = [65.5, 78.2, 54.8, 92.1, 70.0]

srednia = sum(wagi) / len(wagi)

kwadraty = []
for waga in wagi:
    kwadrat = (waga - srednia)**2
    kwadraty.append(kwadrat)


wariancja = sum(kwadraty) / len(kwadraty)
odchylenie_standardowe = wariancja**0.5

print(f"Średnia: {srednia}")
print(f"Odchylenie standardowe: {odchylenie_standardowe:.4}")
print('='*45)
print()
print('-'*45)
print(F"Wartość Oryginalna | Z-score | Interpretacja")
print('-'*45)

lista_z = []
for waga in wagi:
    z_score = (waga -srednia) / odchylenie_standardowe
    if z_score > 2:
        interpretacja = 'wartość bardzo wysoka'
    elif abs(z_score) < 1:
        interpretacja = 'wartość typowa'
    elif z_score < -2:
        interpretacja = 'wartość bardzoniska'
        
    lista_z.append(z_score)
    print(f"{waga:^19}| {round(z_score, 2):^8}| {interpretacja}")

print('-'*45)
print()
srednia_z = sum(lista_z) / len(lista_z)
kwadraty_z = [(z -srednia_z)**2 for z in lista_z]
wariancja_z = sum(kwadraty_z) / len(kwadraty_z)
odchylenie_standardowe_z = wariancja_z**0.5

print(f"Średnia_z: {srednia_z}")
print(f"Odchylenie standardowe_z: {odchylenie_standardowe_z:.4}")
```
